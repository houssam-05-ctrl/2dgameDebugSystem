````python
# Aanalyse d'une image
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from google.colab.patches import cv2_imshow # Indispensable dans Colab

# 1. Uploader une image (ou utilisez une image déjà présente)
print("Veuillez uploader une image (ex: image1.jpg) :")
uploaded = files.upload()
image_path = list(uploaded.keys())[0]

# 2. Lire l'image
img = cv2.imread(image_path)
if img is None:
    raise FileNotFoundError("L'image n'a pas pu être chargée. Vérifiez le nom du fichier.")

# 3. Afficher l'image originale (cv2_imshow attend du BGR)
print("Image originale :")
cv2_imshow(img)

# 4. Convertir en niveaux de gris
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print("\nImage en niveaux de gris :")
cv2_imshow(gray)

# 5. Détection de contours (Canny)
edges = cv2.Canny(gray, 100, 200) # Les seuils peuvent être ajustés
print("\nContours détectés (Canny) :")
cv2_imshow(edges)
````

***1. Que représente la matrice retournée par OpenCV pour l'image ?***

OpenCV retourne une matrice NumPy tridimensionnelle de forme (hauteur, largeur, 3) pour une image couleur. Les trois canaux sont dans l'ordre BGR (Bleu, Vert, Rouge), contrairement à l'ordre RGB utilisé par la plupart des autres bibliothèques. Chaque valeur est un entier non signé sur 8 bits (0-255). Pour une image en niveaux de gris, la matrice est bidimensionnelle (hauteur, largeur).

***2. Pourquoi convertit-on souvent l'image en niveaux de gris avant traitement ?***

La conversion en niveaux de gris réduit la quantité de données à traiter (un seul canal au lieu de trois), ce qui accélère les calculs. De plus, de nombreux algorithmes (comme Canny, les détecteurs de visages de Dlib, etc.) se basent sur l'intensité lumineuse et non sur la couleur. Cela permet aussi de s'affranchir des variations de couleurs qui peuvent être sources de bruit.

***3. Que peut nous apporter le filtre Gaussien ?***

cv2.GaussianBlur(gray, (5, 5), 0) applique un lissage gaussien qui réduit le bruit haute fréquence (bruit aléatoire) de l'image. C'est une étape de prétraitement essentielle avant la détection de contours (Canny) pour éviter que le bruit ne soit interprété comme de faux contours. Le noyau (5,5) définit la taille de la zone de flou ; plus il est grand, plus le flou est prononcé.

````python
# Cellule 3 : Exercice 2 - Webcam dans Colab (via JavaScript)
import cv2
import numpy as np
from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode
from google.colab.patches import cv2_imshow
import PIL.Image
import io

# --- Code JavaScript pour capturer une frame de la webcam ---
def take_photo(filename='photo.jpg', quality=0.8):
    js = Javascript('''
    async function takePhoto(quality) {
      const div = document.createElement('div');
      const capture = document.createElement('button');
      capture.textContent = '📸 Capturer une frame';
      div.appendChild(capture);

      const video = document.createElement('video');
      video.style.display = 'block';
      const stream = await navigator.mediaDevices.getUserMedia({video: true});

      document.body.appendChild(div);
      div.appendChild(video);
      video.srcObject = stream;
      await video.play();

      // Attendre un peu que la vidéo se stabilise
      await new Promise(resolve => setTimeout(resolve, 1000));

      const canvas = document.createElement('canvas');
      canvas.width = video.videoWidth;
      canvas.height = video.videoHeight;
      canvas.getContext('2d').drawImage(video, 0, 0);

      // Arrêter la caméra
      stream.getVideoTracks()[0].stop();
      div.remove();
      return canvas.toDataURL('image/jpeg', quality);
    }
    ''')
    display(js)
    data = eval_js('takePhoto({})'.format(quality))
    binary = b64decode(data.split(',')[1])
    return binary

# --- Utilisation : capturer une frame, la convertir et la traiter ---
try:
    print("Cliquez sur le bouton 'Capturer une frame' qui vient d'apparaître.")
    binary = take_photo()
    # Convertir les données binaires en image OpenCV
    image = np.frombuffer(binary, dtype=np.uint8)
    frame = cv2.imdecode(image, cv2.IMREAD_COLOR)

    # Appliquer le filtre niveaux de gris en temps réel (ici sur la frame capturée)
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    print("Frame originale :")
    cv2_imshow(frame)
    print("Frame en niveaux de gris :")
    cv2_imshow(gray_frame)

except Exception as e:
    print(f"Erreur : {e}. Assurez-vous d'autoriser l'accès à la caméra dans votre navigateur.")
````

***Que se passe-t-il si vous utilisez une vidéo enregistrée à la place de la webcam ?***

## Réponses aux questions — Exercice 6 (Landmarks)

***1. Quelles parties du visage sont retournées par `face_landmarks` ?***

La fonction retourne un dictionnaire avec les clés suivantes :
`chin` (menton), `left_eyebrow`, `right_eyebrow`, `nose_bridge`, `nose_tip`,
`left_eye`, `right_eye`, `top_lip`, `bottom_lip`.
Chaque clé contient une liste de coordonnées `(x, y)` des points correspondants (68 points au total).

***2. Calculez le rapport d'ouverture de l'œil (EAR) et comment l'utiliser pour la somnolence ?***

L'EAR (*Eye Aspect Ratio*) mesure l'ouverture de l'œil. La formule est :

$$EAR = \frac{||p_2 - p_6|| + ||p_3 - p_5||}{2 \cdot ||p_1 - p_4||}$$

où $p_1$ à $p_6$ sont les six points de l'œil dans l'ordre.

Pour détecter la somnolence, on calcule l'EAR en continu sur une vidéo.
Si l'EAR descend en dessous d'un seuil (ex: 0.25) pendant un certain nombre
de frames consécutives (correspondant à ~2 secondes), on considère que la personne
ferme les yeux de manière prolongée et on déclenche une alarme.
C'est la base du mini-projet "Détecteur de somnolence".

## Réponses aux questions — Exercice 7 (YOLO vs Mask R-CNN)

***1. Comparez le temps d'inférence de YOLOv8n et de Mask R-CNN sur la même image.***

YOLOv8n (nano) est conçu pour être extrêmement rapide. Sur le GPU T4 de Colab,
son inférence prend généralement **moins de 50 ms** par image.
Mask R-CNN, avec son architecture plus lourde (ResNet-50 + FPN), prend plutôt
**de l'ordre de 200 à 500 ms**. YOLO est donc significativement plus rapide,
ce qui le rend adapté au temps réel.

***2. YOLO fournit-il des masques de segmentation ? Testez `yolov8n-seg.pt`.***

Le modèle `yolov8n.pt` standard ne fournit que des **boîtes englobantes** (*bounding boxes*).
Pour obtenir des masques de segmentation, il faut utiliser un modèle suffixé par `-seg`,
comme `yolov8n-seg.pt` :

```python
model_seg = YOLO("yolov8n-seg.pt")
results_seg = model_seg(img_path)
cv2_imshow(results_seg[0].plot())
```

Ce modèle fournit des masques polygonaux pour chaque instance détectée.

***3. Dans quels cas préférer chaque modèle ?***

- **Mask R-CNN** : À préférer lorsque la **précision** est primordiale et que le temps
  de calcul n'est pas critique (analyse d'images médicales, forensique, traitement par lots).
  Il fournit des masques de segmentation très précis.

- **YOLO (v8)** : À préférer pour les applications **temps réel** (détection sur webcam,
  véhicules autonomes, robotique) où la vitesse est essentielle.
  YOLOv8-seg offre un excellent compromis vitesse/précision pour la segmentation.

#Mini-projet : Floutage automatique des visages (Anonymisation)

**Objectif :** Détecter les visages sur une image et appliquer un flou gaussien (ou une pixelisation) sur chaque visage.
Ceci est une application concrète de la protection de la vie privée (conformité RGPD) avant de publier une image sur internet.

````python
# Mini-projet : Floutage automatique des visages (Version Image)
import cv2
import dlib
from google.colab.patches import cv2_imshow
from google.colab import files

# 1. Charger l'image
print("Uploadez une image contenant des visages (ex: people.jpg) :")
uploaded = files.upload()
img_path = list(uploaded.keys())[0]

img = cv2.imread(img_path)
assert img is not None, "Image introuvable."

# 2. Initialiser le détecteur de visages Dlib
detector = dlib.get_frontal_face_detector()
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 3. Détection des visages
faces = detector(gray)
print(f"Nombre de visages détectés : {len(faces)}")

# 4. Appliquer le flou sur chaque visage
# On travaille sur une copie pour ne pas altérer l'original si besoin
img_blurred = img.copy()

for face in faces:
    x, y, w, h = face.left(), face.top(), face.width(), face.height()

    # S'assurer que les coordonnées sont dans les limites de l'image
    x, y = max(0, x), max(0, y)
    w = min(w, img.shape[1] - x)
    h = min(h, img.shape[0] - y)

    # Extraire la région du visage (ROI)
    roi = img[y:y+h, x:x+w]

    # Appliquer un flou gaussien très fort (noyau impair et grand)
    # (99, 99) donne un effet "anonymisé"
    roi_blurred = cv2.GaussianBlur(roi, (99, 99), 30)

    # Remplacer la zone du visage par la version floutée
    img_blurred[y:y+h, x:x+w] = roi_blurred

# 5. Affichage du résultat
print("\nImage originale :")
cv2_imshow(img)

print("\nImage avec visages anonymisés :")
cv2_imshow(img_blurred)
````

````python
# Bonus : Floutage en temps réel sur une vidéo enregistrée
import cv2
import dlib
from google.colab.patches import cv2_imshow
from google.colab import files
import os

# Uploader une courte vidéo (ex: video.mp4)
print("Uploadez une courte vidéo (mp4) :")
uploaded = files.upload()
video_path = list(uploaded.keys())[0]

detector = dlib.get_frontal_face_detector()
cap = cv2.VideoCapture(video_path)
frame_count = 0

while True:
    ret, frame = cap.read()
    if not ret:
        break

    frame_count += 1
    # Traiter une frame sur 5 pour éviter de faire ramer Colab
    if frame_count % 5 != 0:
        continue

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces = detector(gray)

    for face in faces:
        x, y, w, h = face.left(), face.top(), face.width(), face.height()
        x, y = max(0, x), max(0, y)
        w, h = max(1, w), max(1, h) # éviter w=0

        roi = frame[y:y+h, x:x+w]
        roi_blurred = cv2.GaussianBlur(roi, (51, 51), 0)
        frame[y:y+h, x:x+w] = roi_blurred

    # Afficher seulement quelques frames pour ne pas faire exploser la mémoire
    if frame_count % 30 == 0:
        cv2_imshow(frame)

cap.release()
print("Fin de la vidéo.")
````
